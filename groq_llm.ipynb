# Alibaba Cloud - qwen

```python
from groq import Groq
os.environ["GROQ_API_KEY"] = "Give Grop Api Key"

client = Groq()
completion = client.chat.completions.create(
    model="qwen/qwen3-32b",
    messages=[
      {
        "role": "user",
        "content": "how are you"
      }
    ],
    temperature=0.6,
    max_completion_tokens=4096,
    top_p=0.95,
    reasoning_effort="default",
    stream=True,
    stop=None
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
```

# Groq

```python
from groq import Groq

client = Groq()
completion = client.chat.completions.create(
    model="groq/compound",
    messages=[
      {
        "role": "user",
        "content": "campare computer and computer science "
      }
    ],
    temperature=1,
    max_completion_tokens=1024,
    top_p=1,
    stream=True,
    stop=None,
    compound_custom={"tools":{"enabled_tools":["web_search","code_interpreter","visit_website"]}}
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
```

```python

import os
from groq import Groq

client = Groq()
filename = "/content/drive/MyDrive/Notebooks/a.mp3"

with open(filename, "rb") as file:
    transcription = client.audio.transcriptions.create(
      file=(filename, file.read()),
      model="whisper-large-v3",
      temperature=0,
      response_format="verbose_json",
    )
    print(transcription.text)
```

```python

import os
from groq import Groq

client = Groq()

filename = "/content/drive/MyDrive/Notebooks/a.mp3"

with open(filename, "rb") as file:
    transcription = client.audio.transcriptions.create(
        file=("audio.m4a", file.read()),  # keep tuple format
        model="whisper-large-v3-turbo",
        temperature=0,
        response_format="verbose_json",
    )

res=transcription.text.split('.')

for i in res:
  print(i)
```
