```python
df_country = pd.read_excel("Country-Code.xlsx")
df_country.head()
```

```python
df.columns
```

```python
final_df=pd.merge(df, df_country, on='Country Code', how='left')
final_df.head(2)
```

```python
# to check datatype
final_df.dtypes
```

```python
final_df.columns
```

```python
final_df.Country.value_counts()
```

```python
country_names = final_df.Country.value_counts().index
country_names
```

```python
country_value=final_df.Country.value_counts().values
country_value
```

```python
plt.pie(x=country_value[:3], labels=country_names[:3], autopct='%1.2f%%' )
plt.show()
```

Observation: Zomato maximum records or transactions are from India after that USA and then UK

```python
final_df.groupby(['Aggregate rating', 'Rating color', 'Rating text']).size()
```

```python
ratings=final_df.groupby(['Aggregate rating', 'Rating color', 'Rating text']).size()\
.reset_index().rename(columns={0:'Rating Count'})
```

```python
ratings
```

## Obervation:
1. Rating is between 4.5 to 4.9 ----> Excellent
2. Rating is between 4.0 to 4.4 ----> very good
3.  Rating is between 3.5 to 3.9 ----> good
4.   Rating is between 3.0 to 3.4 ----> Average
5.    Rating is between 2.5 to 2.9 ----> Average
6. Rating is between 2.0 to 2.4 ----> poor

```python
ratings.head()
```

```python
import matplotlib
matplotlib.rcParams['figure.figsize'] = (12,6)
sns.barplot(x='Aggregate rating', y='Rating Count', data=ratings, palette='winter')
plt.show()
```

```python
sns.barplot(
    x='Aggregate rating', y='Rating Count', 
    data=ratings, hue='Rating color',
    palette=['black', 'red', 'orange', 'yellow', 'green', 'green'] 
)
plt.show()
```

```python
# count plot
sns.countplot(
    x='Rating color', data=ratings,
    palette=['black', 'red', 'orange', 'yellow', 'green', 'green']
)
plt.show()
```
