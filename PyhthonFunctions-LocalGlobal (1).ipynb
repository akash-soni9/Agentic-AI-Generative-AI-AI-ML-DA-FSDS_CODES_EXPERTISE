# **Pyhthon Functions & Local, Global variable**

## 1. Positional arguments:

```python
def add(a, b):
    print(a + b)
add(10, 5)
```

```python
def multiply(a, b):
    return a * b
multiply(10, 5)
```

```python
def full_name(first, last):
    print(first, last)
full_name('Akash', 'Soni')
```

```python
def avg(a, b, c):
    average = (a+b+c)/3
    print(round(average, 2))
avg(10, 3, 54)
```

```python
def power(base, exp):
    print(base **  exp)
power(3, 2)
power(3,3)
```

## 2. Default Arguments

```python
def greet(name='Guest'):
    print('Hello,', name)
    
greet()
greet('Akash')
```

```python
def country(city, nation = 'India'):
    print(city, 'is in', nation)
country('Ranchi')
```

```python
def employee(name, salary = 25000):
    print(name, salary)
employee("xyzz")
```

```python
def bonus(amount, bonus=1000):
    print(amount + bonus)
bonus(23000)
```

```python
def ticket_price(person = 'Adult', price = 120):
    print(person, ':', price)
ticket_price()
ticket_price('Kid', 50)
```

## 3. Keyword Arguments

```python
def student(name,  age):
    print(name, age)
student(age=21, name='Akash')
student(name = 'Akash', age = 21)
```

```python
def product(name, price):
    print(name, price)
product(price = 450, name='shoes')
```

```python
def login(username, password):
    print('user:', username)
login(password = '1232', username='Akash')
```

```python
def profile(name, city, country):
    print(name, city, country)
profile(city='Ranchi', name='Abhishek', country='India')
```

```python
def rectangle(l,w):
    print('Area', l*w)
rectangle(w=4, l=8)
```

## 4. Variable-Length Arguments (*args)

```python
def total(*number):
    print(sum(number))
total(10, 20, 30)
```

```python
def subjects(*sub):
    print(sub)
subjects('Math', 'science', 'English')
```

```python
def max_num(*values):
    maximum = max(values)
    print(maximum)
max_num(12,232,32,345, 34)
```

```python
def marks(*m):
    for i in m:
        print(i)
marks(78,56,90,45)
```

```python
def arithmetics(*nums):
    add = 0
    mul = 1
    for i in nums:
        add += i
        mul *= i
    return add, mul
addition, multiplication = arithmetics(3,2,4,2)
print('addition:',addition)
print('multiplication:',multiplication)
    
```

## 5. Variable-Length Keyword Arguments `**kwargs`

```python
def person(**data):
    print(data)
person(name='Akash', age=22, city = 'Ranchi')
```

```python
def car(**info):
    for k, v in info.items():
        print(k, ':', v)
car(model = 'xuv700', color='black', price=2300000)
```

```python
def student_info(**detail):
    print(detail)
    
student_info(rollnp=7, name='Mahindra', marks=100)
```

```python
def product_detail(**items):
    
    for k, v in items.items():
        print(k, '=', v)
        
product_detail(name = 'laptop', price = 40000, brand='Hp')
```

```python
def order(**O):
    for key, value in O.items():
        print(key, value)
order(item ='Pizza', size='Regular', quantity=2)
```

## 6. Local & Global Variables

```python
def show():
    x = 10
    print(x)
show()
```

```python
y = 50
def display():
    print(y)
display()
print(y)
```

```python
count = 5

def update():
    global count
    count += 5
update()
print(count)
```

```python
def test():
    msg = 'Hello'
    print(msg)
test()
```

```python
name = 'Akash'

def greeting():
    local_name = 'Guest'
    print('hi', local_name)
    print('welcome', name)
greeting()
```

## Real-world example
### 1. Online Shopping — Calculate Total Bill

```python
def calculate_total(price, quantity, discount=0.10, gst = 0.18):
    subtotal = price * quantity
    subtotal -= subtotal * discount
    total = subtotal + (subtotal * gst)

    return total
    
print('Final Amount:', calculate_total(1200, 2))
```

### 2. Food Delivery App (Swiggy/Zomato) — Delivery Fare

```python
def delivery_charge(distance_km, base_rate = 10, surge = 1.2):
    return distance_km * base_rate * surge

print('Delivery Fare:', delivery_charge(5))
```

### 3. Attendance / Employee System — Punch IN Tracking

```python
def attendance_status(hours_worked):
    if hours_worked >= 8:
        return 'Present'
    return 'Half Day'
print(attendance_status(7.5))
```

### 4. Weather App — Convert Temperature

```python
def to_fehrenheit(celcius):
    return (celcius * 9/5) + 32

print('Temperature in °F:', to_fehrenheit(30))
```

### 5. Banking System — Withdraw & Update Balance

```python
balance = 10000  # global variable

def withdraw(amount):
    global balance

    if amount <= balance:
        balance -= amount
        return f'Withdraw: ₹{amount}, Remaining: ₹{balance}'
    else:
        return 'Insuficient Balance'
print(withdraw(3000))
print(withdraw(8000))
```
