```python
def count(lst):
    even = 0
    odd = 0
    for i in lst:
        if i % 2==0:
            even += 1
        else:
            odd += 1
    return even, odd
    
lst = [2,45,64,23,67,34, 7]
even, odd=count(lst)

print('even: ',even)
print('odd: ',odd)

print('Even Number: {} and odd Number: {}'.format(even, odd))
```

**fibonacci series**

```python
def fib(n):
    a = 0
    b = 1
    print(a, end=' ')
    print(b, end=' ')

    for i in range(0, n):
        c = a + b
        a = b
        b = c

        print(c, end=' ')

fib(10)
```

## recursion concept:
factorial of a number

```python
def factorial(n):
    if (n == 0 or n == 1):
        return 1
    return n * factorial(n-1)
factorial(0)
```

find greatest number amongs the 3 number

```python
def greatest(a, b, c):
    if (a>b and a>c):
        return a
    elif (b > a and b > c):
        return b
    else:
        return c

greatest(a=23, b=4, c=34)        
```

```python
def rem(l, word):
    n = []
    for item in l:
        if not (item == word):
            n.append(item.strip(word))
    return n

l = ['python' , 'java', 'Jython', 'Php', 'Sql', 'dart']
print(rem(l, 'on'))
```

```python
# print list of element using recursion
def show(List, idx=0):
    if idx == len(List):
        return
    print(List[idx], end=" ")
    show(List, idx+1)

fruits = ['a', 'v', 'r', 't']
show(fruits)

```
