# Simple Linear Regression On Housing Data

```python
# Importing Libraries
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# prevent NumPy from truncating large arrays while printing
np.set_printoptions(threshold = sys.maxsize)

# Loading dataset
url = r'C:\Users\akash\OneDrive\Desktop\FSDS\ML\SLR\House price prediction\House_data.csv'
dataset = pd.read_csv(url)

#Dividing the dataset into x and y
space = dataset['sqft_living']
price = dataset['price']

x = np.array(space).reshape(-1, 1)
y = np.array(price)

# Spliting the data into train and test
xtrain, xtest, ytrain, ytest = train_test_split(
    x, y, test_size=1/3, random_state=42
)

# Now train the model with xtrain and ytrain
regressor = LinearRegression()
regressor.fit(xtrain, ytrain)

# Predict the prices
pridiction = regressor.predict(xtest)

# Visualizing the Training test result
plt.scatter(xtrain, ytrain, color='red')
plt.plot(xtrain, regressor.predict(xtrain), color = 'blue')
plt.title('Visualize for Training Dataset')
plt.xlabel('space')
plt.ylabel('price')
plt.show()

#Visualize the test result
plt.scatter(xtest, ytest, color='red')
plt.plot(xtrain, regressor.predict(xtrain), color='blue')
plt.title('Visualize for Test DataSet')
plt.xlabel("Space")
plt.ylabel("Price")
plt.show()


bias = regressor.score(xtrain, ytrain)
print(bias)

variance = regressor.score(xtest, ytest)
print(variance)
```
