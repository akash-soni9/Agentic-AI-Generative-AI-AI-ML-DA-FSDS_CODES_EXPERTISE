# **Recursion and lambda function**

# Recursion:
Recursion is a programming technique where a function calls itself to solve 

a smaller instance of the same problem until it reaches a base condition (stopping point).


### Why Use Recursion?

To break complex problems into simpler subproblems.

Useful for tasks like factorial, Fibonacci series, tree/graph traversal, tower of Hanoi, etc.

### Structure of a Recursive Function

1. Every recursive function has:

2. Base case – Stops recursion.

Recursive call – Calls itself with a smaller input.

```python
def wish():
    print('hello')
    print('hi')
wish()  # this not a recursion
```

```python
def wish():
    print('Hello')
    print('Hi')
    wish()  # this is a recursion becoz func call itself within func
wish() # but there is no base condition to stop the program at point
```

## Factorial using Recursion

```python
def fact(n):
    if (n == 0 or n == 1): # base condition
        return 1
    return n * fact(n -1)

result = fact(5)
print(result)
```

## Fibonacci Series using Recursion

```python
def fibonacci(n):
    if n <= 1:  # Base case
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(6))
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

def print_series(num):
    for i in range(num):
        print(fibonacci(i), end=" ")

print_series(7)
```

## Anonymous function | lambda function

```python
def square(a):   # this is not lambda function
    return a * a # we can do this using lambda function
square(5)
```

```python
f = lambda a : a * a # lambda function single line function, it is very fast 
square = f(5)
print(square)
```

```python
add = lambda a, b : a + b
sub = lambda a, b : a - b

subtract = sub(4,1)

print(add(4,1))
print(subtract)
```

```python
f = lambda a, b : a + b
f1 = lambda a, b : a - b
f2 = lambda a, b : a * b

add = f(4, 1)
sub = f1(4, 1)
mul = f2(4, 2)

print(add)
print(sub)
print(mul)
```

## `filter()`

```python
def is_even(n):
    return n % 2 == 0

nums = [3,2,6,8,4,6,2,9]

even = list(filter(is_even, nums))
print(even)
```

```python
def is_odd(n):
    return n % 2 != 0

nums = [3,2,6,8,4,6,2,9]

odd = list(filter(is_odd, nums))
print(odd)
```

```python
nums = [3,2,6,8,4,6,2,9]

evens = list(filter(lambda n : n%2 == 0, nums))
print(evens)
```

```python
nums = [3,2,6,8,4,6,2,9]

odds = list(filter(lambda n : n%2 != 0, nums))
print(odds)
```

```python
nums = [3,2,6,8,4,6,2,9, 34,77, 120]

even = list(filter(lambda n : n%2 == 0, nums))
odd = list(filter(lambda n : n%2 != 0, nums))

print(even)
print(odd)
```

## `map()`
- What ever even number I have from the assigned list
- I want to double the even number i.e 2 become 4  || 4 become 6 || 6 become 8
- that we will do using map function 
- this largly we are using in google map reduce programme
- we can build using user define & lambda

```python
def update(n):
    return n+2
nums = [3,2,6,8,4,6,2,9]

evens = list(filter(lambda n : n%2 == 0, nums))
double = list(map(update, evens))

print(evens)
print(double)
```

```python
nums = [3,2,6,8,4,6,2,9]

evens = list(filter(is_even, nums))
double = list(map(lambda n: n*2, evens))

print(evens)
print(double)
```

```python
nums = [3,2,6,8,4,6,2,9]

evens = list(filter(lambda n : n%2 == 0, nums))

double = list(map(lambda n : n*2, evens))
double_ = list(map(lambda n : n + 2, evens))
double_l = list(map(lambda n : n - 2, evens))

print(evens)

print(double)
print(double_)
print(double_l)
```

## `reduce()`
- i want to perform reduce now 
- i want reduce all the values 
- reduce you can add only 2 values
- [4, 12, 16, 8, 12, 4] if you sum everything then you will get 56

```python
from functools import reduce 

nums = [3,2,6,8,4,6,2,9]

evens = list(filter(is_even, nums))

double = list(map(lambda n : n*2, evens))
sums = reduce(lambda a, b : a + b, double)

print(evens)
print(double)
print(sums)
```

## `Python Decorators(@)`:

- Python decorator is nothing but a function which takes another function as argument and return a function

- it adds extra functionality to another function without modifying its actual code.

```python
def div(a, b):  # this is not decorator
    print(a / b)
div(4, 2)
```

```python
def div(a, b):
    if a<b:
        a, b = b, a
    print(a / b)
div(2, 4)
```

```python
# lets write real code of decorator
```

```python
def div(a, b):
    print(a / b)

def div_decorator(func):
    def inner(a, b):
        if a < b:
            a, b = b, a
            return func(a, b)
    return inner

div = div_decorator(div) # this is decorator 
div(2, 4)
```

```python
def my_decorator(func):
    def wrapper():
        print('Something is happening before the function is called')
        #func()
        print('Something is happening after the function is called')
    return wrapper


@my_decorator
def say_hello():
    print('Hello')

say_hello()
```

```python
def my_decorator(func):
    def wrapper():
        print('Something is happening before the function is called')
        func()
        print('Something is happening after the function is called')
    return wrapper


@my_decorator
def say_hello():
    print('Hello')

say_hello()
```

```python
import time

def timer(func):
    def wrapper():
        start = time.time()
        func()
        end = time.time()
        print(f"Execution Time: {end - start} seconds")
    return wrapper

@timer
def test_function():
    for _ in range(1000000):
        pass

test_function()
```

```python
def repeat(n):
    def decorator(func):
        def wrapper():
            for _ in range(n):
                func()
        return wrapper
    return decorator

@repeat(3)
def greet():
    print("Hello!")

greet()
```
