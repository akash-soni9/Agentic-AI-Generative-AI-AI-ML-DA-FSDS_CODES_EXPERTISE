# **Online Shopping Billing System (Console App)**

```python
PRODUCTS = {
    1: {'name': 'Laptop', 'price': 55000},
    2: {'name': 'Headphones', 'price': 1500},
    3: {'name': 'Smartphone', 'price': 25000},
    4: {'name': 'Keyboard', 'price': 800},
    5: {'name': 'Mouse', 'price': 500},
}

cart = []   # Global cart

def add_to_cart(product_id, quantity=1):
    if product_id not in PRODUCTS:
        print("❌ Invalid Product ID")
        return

    product = PRODUCTS[product_id]
    cart.append({
        "product_id": product_id,
        "name": product["name"],
        "qty": quantity,
        "price": product["price"]
    })
    print(f"Added {quantity} x {product['name']} to cart")


def show_product_details(*, product_id=None):
    if product_id is None or product_id not in PRODUCTS:
        print("❌ Please provide a valid Product ID")
        return
    product = PRODUCTS[product_id]
    print(f"Product: {product['name']} | Price: ₹{product['price']}")


def add_multiple_to_cart(*product_ids):
    for pid in product_ids:
        add_to_cart(pid)


def apply_discount(total_amount, **discounts):
    total_discount = sum(discounts.values())
    final_amount = max(total_amount - total_discount, 0)
    return final_amount, total_discount


GST_RATE = 0.18   # Global


def calculate_bill():
    if not cart:
        print("🛒 Cart is empty! Add items first.")
        return None

    subtotal = 0
    print("\n----- CART ITEMS -----")
    for item in cart:
        line_total = item["qty"] * item["price"]
        subtotal += line_total
        print(f"{item['name']} (x{item['qty']}) = ₹{line_total}")

    gst_amount = subtotal * GST_RATE
    total_with_gst = subtotal + gst_amount

    print("\nSubtotal: ₹", subtotal)
    print("GST @", int(GST_RATE * 100), "%: ₹", round(gst_amount, 2))
    print("Total (incl. GST): ₹", round(total_with_gst, 2))

    return subtotal, gst_amount, total_with_gst


def print_invoice(customer_name, *notes, **extra_info):
    print("\n========= INVOICE ==========")
    print("Customer:", customer_name)

    if extra_info:
        for k, v in extra_info.items():
            print(f"{k.capitalize()}: {v}")

    print("-----------------------------")
    result = calculate_bill()
    if result is None:
        return

    subtotal, gst_amount, total = result

    print("\n Applying discount (if any)...")
    final_amount, total_discount = apply_discount(
        total,
        festival=extra_info.get("festival_discount", 0),
        coupon=extra_info.get("coupon_discount", 0)
    )

    print("Total Discount: ₹", total_discount)
    print("Amount Payable: ₹", round(final_amount, 2))

    if notes:
        print("\nNotes:")
        for n in notes:
            print("-", n)

    print("================================")
    print("Thank you for shopping with us! 😊")


def show_product():
    print("\n AVAILABLE PRODUCTS:")
    for pid, p in PRODUCTS.items():
        print(f"{pid}. {p['name']} - ₹{p['price']}")


def main():
    print("🛍️ Welcome to our Shopping Mart")

    while True:
        show_product()
        choice = input(
            "\n1. Add to cart\n"
            "2. Add multiple items\n"
            "3. Show product details\n"
            "4. Checkout\n"
            "5. Exit\n"
            "Enter choice: "
        )

        if choice == "1":
            pid = int(input("Enter Product ID: "))
            qty_str = input("Enter Quantity (default 1): ")
            qty = int(qty_str) if qty_str.strip() != "" else 1
            add_to_cart(pid, qty)

        elif choice == "2":
            ids = input("Enter Product IDs separated by comma (e.g. 1,2,3): ")
            id_list = [int(i.strip()) for i in ids.split(",") if i.strip().isdigit()]
            add_multiple_to_cart(*id_list)

        elif choice == "3":
            pid = int(input("Enter Product ID to view details: "))
            show_product_details(product_id=pid)

        elif choice == "4":
            name = input("Enter Customer Name: ")
            payment = input("Payment Mode (UPI/Cash/Card): ")
            city = input("City: ")
            festival_disc_str = input("Festival Discount (₹, 0 if none): ")
            coupon_disc_str = input("Coupon Discount (₹, 0 if none): ")

            festival_disc = int(festival_disc_str) if festival_disc_str.strip() != "" else 0
            coupon_disc = int(coupon_disc_str) if coupon_disc_str.strip() != "" else 0

            print_invoice(
                name,
                "Keep your invoice for future reference.",
                "Customer support: support@pythonshop.com",
                payment_mode=payment,
                city=city,
                festival_discount=festival_disc,
                coupon_discount=coupon_disc
            )
            break

        elif choice == "5":
            print("Exiting... Goodbye! ")
            break

        else:
            print(" Invalid choice, try again.")


if __name__ == "__main__":
    main()
```
